```python
import matplotlib.pyplot as plt

plt.hist(df['tenure'], bins=20)
plt.xlabel('Antigüedad (meses)')
plt.ylabel('Cantidad de clientes')
plt.title('Distribución de la antigüedad de los clientes')
plt.show()
```

```python
for churn in ["No", "Yes"]:
    plt.hist(
        df.loc[df["Churn"] == churn, "tenure"],
        bins=20,
        alpha=0.5,
        label=f"Churn: {churn}"
    )

plt.xlabel("Antigüedad (meses)")
plt.ylabel("Cantidad de clientes")
plt.title("Antigüedad de los clientes según abandono")
plt.legend()
plt.show()
```

```python
tasa_churn = df.groupby("tenure")["Churn"].apply(
    lambda x: (x == "Yes").mean() * 100
)

tasa_churn.head(15)
```

```python

resumen_tenure = df.groupby("tenure").agg(
    total_clientes=("Churn", "size"),
    clientes_abandonaron=("Churn", lambda x: (x == "Yes").sum()),
    tasa_abandono=("Churn", lambda x: (x == "Yes").mean() * 100)
)

resumen_tenure.head(15).round(2)
```

```python

plt.figure(figsize=(12, 5))

plt.plot(
    resumen_tenure.index,
    resumen_tenure["tasa_abandono"],
    marker="o",
    markersize=4
)

plt.xlabel("Antigüedad del cliente (meses)")
plt.ylabel("Tasa de abandono (%)")
plt.title("Tasa de abandono según antigüedad del cliente")
plt.grid(alpha=0.3)
plt.xticks(range(0, 73, 6))
plt.show()
```

```python

resumen_contrato = df.groupby("Contract").agg(
    total_clientes=("Churn", "size"),
    clientes_abandonaron=(
        "Churn",
        lambda x: (x == "Yes").sum()
    ),
    tasa_abandono=(
        "Churn",
        lambda x: (x == "Yes").mean() * 100
    )
)

resumen_contrato.sort_values(
    "tasa_abandono",
    ascending=False
).round(2)
```

```python
plt.figure(figsize=(9, 5))

plt.bar(
    resumen_contrato.index,
    resumen_contrato["tasa_abandono"]
)

plt.xlabel("Tipo de contrato")
plt.ylabel("Tasa de abandono (%)")
plt.title("Tasa de abandono según tipo de contrato")
plt.ylim(0, 50)
plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python

resumen_pago = df.groupby("PaymentMethod").agg(
    total_clientes=("Churn", "size"),
    clientes_abandonaron=(
        "Churn",
        lambda x: (x == "Yes").sum()
    ),
    tasa_abandono=(
        "Churn",
        lambda x: (x == "Yes").mean() * 100
    )
)

resumen_pago.sort_values(
    "tasa_abandono",
    ascending=False
).round(2)
```

```python

resumen_internet = df.groupby("InternetService").agg(
    total_clientes=("Churn", "size"),
    clientes_abandonaron=(
        "Churn",
        lambda x: (x == "Yes").sum()
    ),
    tasa_abandono=(
        "Churn",
        lambda x: (x == "Yes").mean() * 100
    )
)

resumen_internet.sort_values(
    "tasa_abandono",
    ascending=False
).round(2)
```

```python

resumen_cargos = df.groupby("Churn")["MonthlyCharges"].agg(
    ["count", "mean", "median", "std"]
)

resumen_cargos.round(2)
```

```python
datos_cargos = [
    df.loc[df["Churn"] == "No", "MonthlyCharges"],
    df.loc[df["Churn"] == "Yes", "MonthlyCharges"]
]

plt.figure(figsize=(8, 5))

plt.boxplot(
    datos_cargos,
    tick_labels=["Permanecieron", "Abandonaron"]
)

plt.ylabel("Cargos mensuales")
plt.title("Distribución de cargos mensuales según abandono")
plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python

resumen_total = df.groupby("Churn")["TotalCharges"].agg(
    ["count", "mean", "median", "std"]
)

resumen_total.round(2)
```
